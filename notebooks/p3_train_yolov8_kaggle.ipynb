```python
import yaml

DATASET = "/kaggle/input/TEN-DATASET"   # sửa tên dataset
cfg = {"path": DATASET, "train": "images/train", "val": "images/val",
       "test": "images/test", "names": {0: "license_plate"}}
yaml.safe_dump(cfg, open("/kaggle/working/data.yaml", "w"))
```

```python
from ultralytics import YOLO

EPOCHS = 2   # chạy thử trước, ổn thì đổi ~50
model = YOLO("yolov8s.pt")
model.train(data="/kaggle/working/data.yaml", imgsz=640, epochs=EPOCHS, batch=16,
            device=0, project="/kaggle/working/runs", name="plate_v1")
```

```python
best = YOLO("/kaggle/working/runs/plate_v1/weights/best.pt")
m = best.val(data="/kaggle/working/data.yaml", split="val")   # split="test" chỉ ở tuần đánh giá chính thức
print("mAP50:", m.box.map50, "mAP50-95:", m.box.map)
```
